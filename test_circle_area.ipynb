# Circle Area Tests

This notebook tests `circle_area` imported from `circle_area.py`, keeping the function in one place.

For a radius in metres, $A = \pi r^2$ gives the area in square metres. A radius of 5 m gives approximately 78.54 square metres.

The code cell below shows the example and runs five automated tests covering positive, fractional, zero, and negative radii. It uses Python's built-in `unittest` module.

```python
import math
import unittest

from circle_area import circle_area


radius_metres = 5
area = circle_area(radius_metres)
print(f"Area of a circle with radius {radius_metres} m: {area:.2f} m^2")


class TestCircleArea(unittest.TestCase):
    def test_five_metre_radius(self):
        self.assertAlmostEqual(circle_area(5), 25 * math.pi)

    def test_unit_radius(self):
        self.assertAlmostEqual(circle_area(1), math.pi)

    def test_fractional_radius(self):
        self.assertAlmostEqual(circle_area(0.5), 0.25 * math.pi)

    def test_zero_radius(self):
        self.assertEqual(circle_area(0), 0)

    def test_negative_radius(self):
        with self.assertRaises(ValueError):
            circle_area(-1)


suite = unittest.defaultTestLoader.loadTestsFromTestCase(TestCircleArea)
result = unittest.TextTestRunner(verbosity=2).run(suite)
assert result.wasSuccessful(), "Circle area tests failed."
```
